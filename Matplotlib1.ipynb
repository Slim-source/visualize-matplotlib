```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([80,85,90,95,100])
y=np.array([280,290,300,310,320])
plt.title("Sports watch data")
plt.xlabel("Average pulse")
plt.ylabel("Calories Burnage")
plt.plot(x,y)
# plt.grid()
plt.grid(color='green',linewidth='0.5',linestyle='--')
plt.show()
```

```python
x=np.array([1,2,3,4,5])
y=np.array([3,8,1,10,7])
plt.subplot(1,2,1)
plt.plot(x,y)
plt.title("PLOT 1")
x=np.array([2,4,6,8,10])
y=np.array([6,15,1,2,19])
plt.subplot(1,2,2)
plt.plot(x,y)
plt.title("PLOT 2")
plt.suptitle("SUBPLOT EXAMPLE")
```

```python
y=np.array([35,25,25,15])
myLabels=['Apples','Bananas','Cherries','Dates']
myExplode=[0,0.5,0,0]
plt.pie(y,labels=myLabels,explode=myExplode)
plt.legend(title="FRUITS")
plt.show()
```

```python
# X-coordinates for the 8 data points
x=np.array([7,5,8,3,6,9,1,4])

# Y-coordinates for the 8 data points
y=np.array([99,87,110,86,104,99,88,77])

# Numeric values representing the color mapping for each point
colors=np.array([0,10,20,30,40,50,60,70])

# Sizes in points^2 for each circle representing the data points
sizes=np.array([100,200,300,400,500,600,700,800])

# Create a scatter plot:
# - c=colors: maps the colors array to the colormap
# - s=sizes: sets custom sizes for each individual marker
# - cmap='viridis': uses the 'viridis' color scheme to map numbers to colors
plt.scatter(x,y,c=colors,s=sizes,cmap='viridis')

# Display a colorbar on the side to show the scale mapping of the values
plt.colorbar()

# Render and output the final plot
plt.show()
```
